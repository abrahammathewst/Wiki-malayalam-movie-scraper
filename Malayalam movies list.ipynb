```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
for yr in [2025,2024,2023,2022]:  
    url = 'https://en.wikipedia.org/wiki/List_of_Malayalam_films_of_{}'.format(yr)

    headers = {
        "User-Agent": "Mozilla/5.0 (compatible; research-bot/1.0)"
    }

    page = requests.get(url, headers=headers)

    soup = BeautifulSoup(page.text, 'html.parser')
    
    if yr == 2025:
        table = soup.find_all('table')[4]
        titles = table.find_all('th')
        world_titles = [t.text.strip() for t in titles]
        df = pd.DataFrame(columns=world_titles)
        df.insert(0, "Year", "")
        df.insert(1, "Month", "")
        df.insert(2, "Date", "")
        df = df.drop(columns=["Opening"])
    
    year_to_list = {
    2025: [4, 5, 6, 7],
    2024: [3, 4, 5, 6],
    2023: [3, 4, 5, 6],
    2022: [1, 2, 3, 4]}
    values = year_to_list.get(yr, [])
    
    for i in values:
        table = soup.find_all('table')[i]
        columns = table.find_all('tr')

        for row in columns[1:]:
            row_data = row.find_all('td')
            world_row_data = [data.text.strip() for data in row_data]
            world_row_data.insert(0, yr)
            MONTHS = {
            "JANUARY","FEBRUARY","MARCH","APRIL","MAY","JUNE",
            "JULY","AUGUST","SEPTEMBER","OCTOBER","NOVEMBER","DECEMBER"}
            if world_row_data[1] not in MONTHS:
                world_row_data.insert(1, None)
                if not world_row_data[2].isnumeric():
                    world_row_data.insert(1, None)
            length = len(df)
            df.loc[length] = world_row_data

    df[["Month", "Date"]] = df[["Month", "Date"]].ffill()
    print(yr)
    df.tail()
```

```python
df.drop(columns=["Ref"], inplace=True)
df
```

```python
df.to_csv("Malayalam 2022 to 2025 movies list.csv", index=False)
```
